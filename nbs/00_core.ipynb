The notebooks export into `solid_lens/`, so a module's `__file__` is not the top of the repo. Checkpoints, the character dictionary, and `requirements.txt` stay next to `pixi.toml`. Call `repo_root` instead of assuming where the file lives. In a notebook there is no `__file__`, so the search also starts from the working directory.


```python
#| export
from __future__ import annotations

from pathlib import Path
```

```python
#| export
def repo_root() -> Path:
    "Checkout that holds `models/`, `pixi.toml`, and `requirements.txt`."
    starts = []
    file = globals().get("__file__")
    if file: starts.append(Path(file).resolve().parent)
    starts.append(Path.cwd())
    for start in starts:
        for parent in (start, *start.parents):
            if (parent / "pixi.toml").is_file() and (parent / "requirements.txt").is_file():
                return parent
    raise FileNotFoundError("solid_lens checkout (pixi.toml) was not found")
```

`repo_root` is the directory you clone, not the package directory inside it.


```python
from fastcore.test import test_eq

root = repo_root()
test_eq(root.name, "solid_lens")
test_eq((root / "pixi.toml").is_file(), True)
test_eq((root / "nbs").is_dir(), True)
root.name
```
