`max.nn` does not provide the squeeze-excite hardsigmoid. GELU is the LCNet channel mixer. ReLU is the stem, the DB head, and IntraCL. The cells below need MAX, so `nbdev-test` skips them unless you pass `--flags notest`.


```python
#| export
#| notest
from __future__ import annotations

import numpy as np
from max.graph import TensorValue, ops

from solid_lens.config import DEVICE, DTYPE
```

```python
#| export
#| notest
def hardsigmoid(x: TensorValue, alpha: float, beta: float = 0.5) -> TensorValue:
    zero = ops.constant(np.float32(0), DTYPE, device=DEVICE)
    one = ops.constant(np.float32(1), DTYPE, device=DEVICE)
    return ops.min(one, ops.max(zero, x * np.float32(alpha) + np.float32(beta)))
```

hardsigmoid(x) = min(1, max(0, αx + β)) with β = 0.5. The configs set α to 1/6.


```python
#| export
#| notest
def gelu(x: TensorValue) -> TensorValue:
    return ops.gelu(x)
```

```python
#| export
#| notest
def relu(x: TensorValue) -> TensorValue:
    return ops.relu(x)
```
